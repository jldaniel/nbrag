This notebook consolidates engineering findings and states the business conclusion.


```python
import pandas as pd

project_id = 'MFG-LINEB-2026'
df = pd.read_csv('line_b_yield_MFG-LINEB-2026.csv')

peer_mean = df[df.tool_id != 'Tool_3'].groupby('tool_id')['yield_pct'].mean().mean()
tool_3_mean = df[df.tool_id == 'Tool_3']['yield_pct'].mean()
yield_gap = peer_mean - tool_3_mean

print(f'project_id={project_id}')
print(f'tool_3_mean={tool_3_mean:.2f}')
print(f'peer_tools_mean={peer_mean:.2f}')
print(f'yield_gap_pct_points={yield_gap:.2f}')
```

## Evidence Summary

SPC flagged three out-of-control segments on the individuals chart. Stratified yields show Tool_3 consistently
underperforming peer finishing tools by roughly four percentage points — consistent with tool drift rather than
material lot variation.


```python
evidence = {
    'spc_out_of_control_runs': 3,
    'tool_3_batches': int((df.tool_id == 'Tool_3').sum()),
    'yield_gap_vs_peers': round(yield_gap, 2),
    'manifest_rows': 4800,
}
for key, value in evidence.items():
    print(f'{key}={value}')
```

## Root Cause Determination

Process engineering reviewed Tool_3 calibration logs and correlated the SPC violations with a gradual chuck
pressure drift. Peer tools remained within control limits during the same production window.


```python
primary_yield_driver = 'Tool_3_drift'
recommended_action = 'schedule_tool_3_calibration'
confidence = 'high'

print(f'primary_yield_driver={primary_yield_driver}')
print(f'recommended_action={recommended_action}')
print(f'confidence={confidence}')
print(f'project_id={project_id}')
```
